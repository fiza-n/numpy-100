```python
import matplotlib.pyplot as plt
import numpy as np

v = np.array([1, 2])
w = np.array([3, 1])
result = v + w

fig, ax = plt.subplots()

# v: origin se shuru
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='b', label='v')

# w: v ki tip se shuru (tip-to-tail)
ax.quiver(0,0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='g', label='w')

# result: origin se seedha result tak
ax.quiver(0, 0, result[0], result[1], angles='xy', scale_units='xy', scale=1, color='r', label='v+w')

ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.axhline(1, color='pink', linewidth=1)
ax.axvline(1, color='pink', linewidth=1)
ax.set_aspect('equal')
ax.legend()
plt.grid(True)
plt.show()
```

```python
a = np.array([2,1])
b = np.array([1,3])
c1,c2 = 2,3

result = c1*a + c2*b
print(result)
```

```python
def linear_combination(*args):
    result=[]
    for i in range(len(a)):
        result.append(c1*a[i] + c2*b[i])

    return result

result = linear_combination(2,3,[2,1],[1,3])

print(result)
```
